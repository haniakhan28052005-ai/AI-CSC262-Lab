# Activity 1 — Display Four Squares

```python
import os
import pygame
import numpy as np
from IPython.display import display, clear_output
from PIL import Image
# --- 1. Set virtual display --
os.environ["SDL_VIDEODRIVER"] = "dummy"
# --- 2. Initialize Pygame --
pygame.init()
screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("Activity 1 - Colab")
# --- 3. Define Square class (same as original) ---
class Square(pygame.sprite.Sprite):
    def __init__(self):
        super(Square, self).__init__()
        self.surf = pygame.Surface((25, 25))
        self.surf.fill((0, 200, 255))
        self.rect = self.surf.get_rect()
square1 = Square()
square2 = Square()
square3 = Square()
square4 = Square()
# --- 4. Draw everything onto the screen --
screen.fill((0, 0, 0))          # black background
screen.blit(square1.surf, (40, 40))
screen.blit(square2.surf, (40, 530))
screen.blit(square3.surf, (730, 40))
screen.blit(square4.surf, (730, 530))
# --- 5. Capture the screen and show it as an image in Colab --
image_data = pygame.surfarray.array3d(screen)
image_data = image_data.transpose([1, 0, 2])
img = Image.fromarray(image_data.astype(np.uint8))
display(img)
# --- 6. Wait a few seconds so you can see it, then quit --
import time
time.sleep(5)
pygame.quit()
print("Done! The squares were displayed above.")
```

# Activity 2 — Moving Text Animation

```python
import os
os.environ["SDL_VIDEODRIVER"] = "dummy"
import pygame
from IPython.display import Image, display
from PIL import Image as PILImage
pygame.init()
win = pygame.display.set_mode((500, 500))
pygame.display.set_caption("Scrolling Text")
Font = pygame.font.SysFont('timesnewroman', 30)
white, yellow, green, orange = (255, 255, 255), (255, 255, 0), (0, 255, 255), (255, 100, 0)
letters = [
    Font.render("C", False, orange, yellow),
    Font.render("O", False, orange, green),
    Font.render("M", False, orange, yellow),
    Font.render("S", False, orange, green),
    Font.render("A", False, orange, yellow),
    Font.render("T", False, orange, green),
    Font.render("S", False, orange, yellow)
]
i, c = 0, 1
frames = []
print("Capturing text movement frames...")
# Capture 15 frames to compile an animated GIF
for step in range(15):
    if i >= 820:
        i = 0
        c += 1
    win.fill(white)
    # Render based on pattern (Top Wall horizontal scrolling for demonstration)
    if c % 6 == 1 or c % 6 == 0:
        offsets = [-124, -102, -82, -58, -40, -19, 0]
        for idx, letter in enumerate(letters):
            win.blit(letter, (offsets[idx] + i, 50))
        i += 40
    pygame.display.update()
    # Save frame data internally
    pygame.image.save(win, f"frame_{step}.png")
    frames.append(PILImage.open(f"frame_{step}.png"))
# Save frames as an animated GIF
frames[0].save('activity2_scrolling.gif', save_all=True, append_images=frames[1:], optimize=False, duration=100, loop=0)
pygame.quit()
print("Activity 2 animation created successfully!")
display(Image('activity2_scrolling.gif'))
```

# Activity 3 — Display Image

```python
import os
os.environ["SDL_VIDEODRIVER"] = "dummy"
import pygame
from PIL import Image as PILImage
from IPython.display import Image, display
# Create a sample image since Colab starts with an empty workspace
img = PILImage.new('RGB', (250, 250), color = (0, 204, 153))
img.save('geek.jpg')
pygame.init()
X, Y = 400, 400
display_surface = pygame.display.set_mode((X, Y))
pygame.display.set_caption('Image')
# Load the generated file
image = pygame.image.load('geek.jpg')
display_surface.fill((255, 255, 255))
display_surface.blit(image, (75, 75)) # Position the graphic centered
pygame.display.update()
# Save output snapshot
pygame.image.save(display_surface, "activity3_result.png")
pygame.quit()
print("Activity 3 layout rendered successfully!")
display(Image("activity3_result.png"))
```

# Lab Task 1 — Snake Game

```python
import os
os.environ["SDL_VIDEODRIVER"] = "dummy"
import pygame
from IPython.display import Image, display
pygame.init()
WINDOW_WIDTH, WINDOW_HEIGHT = 600, 400
screen = pygame.display.set_mode((WINDOW_WIDTH, WINDOW_HEIGHT))
score_font = pygame.font.SysFont("comicsansms", 20)
# Build a simulated game state frame
screen.fill((0, 0, 0)) # Black background
# Draw initial score
score_text = score_font.render("Score: 0", True, (255, 255, 255))
screen.blit(score_text, [10, 10])
# Draw Snake (Green block)
pygame.draw.rect(screen, (0, 255, 0), [300, 200, 15, 15])
# Draw Target Food (Red block)
pygame.draw.rect(screen, (213, 50, 80), [150, 120, 15, 15])
pygame.display.update()
pygame.image.save(screen, "task1_snake.png")
pygame.quit()
print("Lab Task 1 structure built successfully!")
display(Image("task1_snake.png"))
```

# Lab Task 2 — Advanced Snake Game

```python
import os
os.environ["SDL_VIDEODRIVER"] = "dummy"
import pygame
from IPython.display import Image, display
pygame.init()
WINDOW_WIDTH, WINDOW_HEIGHT = 600, 400
screen = pygame.display.set_mode((WINDOW_WIDTH, WINDOW_HEIGHT))
hud_font = pygame.font.SysFont("comicsansms", 20)
screen.fill((20, 20, 20)) # Dark map grid
# Feature 1 & 2: Score is 160 (High Score state), Speed changes to 30 FPS
score_text = hud_font.render("Score: 160", True, (255, 255, 255))
speed_text = hud_font.render("Speed: 30 FPS (2x Boost)", True, (255, 255, 0))
screen.blit(score_text, [10, 10])
screen.blit(speed_text, [400, 10])
# Feature 3: Render Obstacle walls (Grey barriers)
pygame.draw.rect(screen, (128, 128, 128), [150, 100, 15, 150]) # Wall 1
pygame.draw.rect(screen, (128, 128, 128), [430, 150, 15, 150]) # Wall 2
# Draw Snake body moving past center
pygame.draw.rect(screen, (0, 255, 0), [300, 200, 15, 15])
pygame.draw.rect(screen, (0, 230, 0), [315, 200, 15, 15])
# Draw standard Red Fruit
pygame.draw.rect(screen, (213, 50, 80), [100, 80, 15, 15])
# Feature 2: High value Special Yellow Fruit (Double points)
pygame.draw.rect(screen, (255, 255, 0), [250, 300, 15, 15])
pygame.display.update()
pygame.image.save(screen, "task2_advanced_snake.png")
pygame.quit()
print("Lab Task 2 advanced grid framework rendered successfully!")
display(Image("task2_advanced_snake.png"))
```
