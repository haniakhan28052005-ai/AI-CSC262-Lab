***6. Add and delete attributes***

```python
class Student:
    def __init__(self, student_id, student_name):
        self.student_id = student_id
        self.student_name = student_name

s = Student(1, "Hania")

# Add attribute
s.student_class = "12th"

print("Before deletion:", s.__dict__)

# Delete attribute
del s.student_name

print("After deletion:", s.__dict__)
```

***7. Display all attributes method***

```python
class Student:
    def __init__(self, student_id, student_name, student_class):
        self.student_id = student_id
        self.student_name = student_name
        self.student_class = student_class

    def display(self):
        for key, value in self.__dict__.items():
            print(key, ":", value)

s = Student(1, "Hania", "12th")
s.display()
```

***9. Memory allocation analysis***

```python
import sys

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

s = Student("Hania", 82)

print("Memory size of object:", sys.getsizeof(s))
print("Memory size of dict:", sys.getsizeof(s.__dict__))
```

***10. Reduce memory usage using __slots__***

```python
class Student:
    __slots__ = ['name', 'marks']

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

s = Student("Hania", 82)

import sys
print("Memory with __slots__:", sys.getsizeof(s))
```

***                               Lab Task 02                              ***

***1. Define classes***

```python
class Trapezoid:
    def __init__(self, a, b, h):
        self.a = a
        self.b = b
        self.h = h

    def area(self):
        return 0.5 * (self.a + self.b) * self.h


class Parallelogram:
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return self.base * self.height
```

***2–5. Comparison, dictionary, iteration***

```python
results = {}

data_sets = [
    (10, 8, 5, 6, 4),
    (12, 6, 7, 5, 3),
    (15, 10, 8, 9, 6)
]

for i, (a, b, h, base, height) in enumerate(data_sets, start=1):
    trap = Trapezoid(a, b, h)
    para = Parallelogram(base, height)

    trap_area = trap.area()
    para_area = para.area()

    results[f"Set {i}"] = {
        "Trapezoid": trap_area,
        "Parallelogram": para_area
    }

# Display results
for key, value in results.items():
    print(f"\n{key}:")
    for shape, area in value.items():
        print(f"{shape} Area = {area}")

# Find maximum
max_shape = ""
max_area = 0

for dataset in results.values():
    for shape, area in dataset.items():
        if area > max_area:
            max_area = area
            max_shape = shape

print("\nShape with maximum area:", max_shape)
print("Maximum area:", max_area)
```
