```python
import pandas as pd
import numpy as np
```

```python
import zipfile

with zipfile.ZipFile("archive (1).zip", 'r') as zip_ref:
    zip_ref.extractall("dataset")

df = pd.read_csv("dataset/Book1.csv")
```

```python
print(df)
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
print(df.describe())
```

```python
passed = df[df["Result"] == "Pass"]
print(passed)
```

```python
failed = df[df["Result"] == "Fail"]
print(failed)
```

```python
avg_marks = df["Marks"].mean()
print("Average Marks:", avg_marks)
```

```python
topper = df[df["Marks"] == df["Marks"].max()]
print(topper)
```

```python
import matplotlib.pyplot as plt

plt.bar(df["Name"], df["Marks"])
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks Analysis")
plt.show()
```
